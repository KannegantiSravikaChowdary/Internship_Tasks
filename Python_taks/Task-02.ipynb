```python
#1.Find the Shortest Word in a Sentence
s="Python makes coding easy"
S=s.split()
shortest=S[0]
for word in S:
    if len(shortest)>len(word):
        shortest=word 
print(shortest)    
```

```python
#2.. Find the Largest Number in a List
num=[12, 45, 7, 89, 23]
largest=num[0]
for i in num:
    if i>largest:
        largest =i
print(largest)        
```

```python
#5.Find the Word with the Maximum Number of Characters
s="Data engineering requires problem solving"
S=s.split()
maximum=(S[0])
for word in S:
    if len(word)>len(maximum):
        maximum=word 
print(maximum)        
```

```python
#6.Count Vowels in a String
s="Education"
vowels="AEIOUaeiou"
c=0
for ch in s:
    if ch in vowels:
        c+=1 
print(c)        
```

```python
#7.Count Consonants in a String
s="Python 123!"
vowels="AEIOUaeiou"
c=0
for ch in s:
    if ch.isalpha() and ch  not in vowels:
        c+=1 
print(c)  
```

```python
#8.Find the Most Frequent Character
s="banana"
freq={}
for ch in s:
    if ch in freq:
        freq[ch]+=1 
    else:
        freq[ch]=1 
max_freq=max(freq.values())
for chs in freq:
    if freq[chs]==max_freq:
        print(chs)
        break

```

```python
#9. Find the Least Frequent Character 
s="successs"
freq={}
for ch in s:
    if ch in freq:
        freq[ch]+=1 
    else:
        freq[ch]=1 
min_freq=min(freq.values())
for chs in freq:
    if freq[chs]==min_freq:
        print(chs)
        break
```

```python
#14. Create a Dictionary of Numbers and Their Squares 
num=[2, 3, 4, 5]
d={}
for nums in num:
    if nums not in d:
        d[nums]=nums**2
print(d)
```

```python
#15.Create a Dictionary of Numbers and Their Even/Odd Status
nums=[1,2,3,4,5]
d={}
for num in nums:
    if num not in d:
        if num%2==0:
            d[num]="Even"
        else:
            d[num]="Odd"
print(d)
        
```

```python
#18.. Find the First Word That Appears Only Once
s="apple banana apple mango banana orange"
S=s.split()
d={}
for word in S:
    if word in d:
        d[word]=d[word]+1 
    else:
        d[word]=1 
unique=[]
for ch in d:
    if d[ch]==1:
        unique.append(ch)
print(unique[0])
```
